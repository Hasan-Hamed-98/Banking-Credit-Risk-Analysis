```python
df.columns = df.columns.str.lower()
```

```python
df.rename(columns={
    'month_duration': 'loan_duration_months',
    'credit_amount': 'loan_amount',
    'n_credits': 'number_of_credits'
}, inplace=True)
```

```python
df.duplicated().sum()
```

```python
df['target_binary'] = df['target'].map({
    'good': 0,
    'bad': 1
})
```

```python
## feature engineering
```

```python
## Feature 1 — Age Group

df['age_group'] = pd.cut(
    df['age'],
    bins=[18,25,35,50,100],
    labels=['18-25','26-35','36-50','50+']
)
```

```python
## Feature 2 — Loan Size Category

df['loan_size'] = pd.qcut(
    df['loan_amount'],
    q=3,
    labels=['Small','Medium','Large']
)
```

```python
## Feature 3 — Risky Employment Flag

df['employment_risk'] = np.where(
    df['years_employment'].isin([
        'unemployed',
        '< 1 year'
    ]),
    'High Risk',
    'Stable'
)
```

```python
## Feature 4 — High Loan Flag

df['high_loan_flag'] = np.where(
    df['loan_amount'] > df['loan_amount'].median(),
    1,
    0
)
```
